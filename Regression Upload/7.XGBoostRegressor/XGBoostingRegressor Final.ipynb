```python
import xgboost
import pandas as pd
dataset = pd.read_csv("insurance_pre.csv")
dataset= pd.get_dummies(dataset, drop_first=True, dtype=int)
independent = dataset [['age', 'bmi', 'children','sex_male', 'smoker_yes']]
dependent = dataset [['charges']]
```

```python
from sklearn.model_selection import GridSearchCV
from xgboost import XGBRegressor
param_grid = {'n_estimators':[50,60], 'learning_rate':[0.01,0.1], 'booster': ['gbtree', 'dart', 'gblinear']}
grid = GridSearchCV(XGBRegressor(), param_grid, refit = True, verbose= 3,n_jobs=-1)
grid.fit(independent, dependent)
print("The R_score value for best parameter {}:".format(grid.best_params_),grid.best_score_)
```
