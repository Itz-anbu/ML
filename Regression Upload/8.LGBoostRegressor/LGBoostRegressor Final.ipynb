```python
import pandas as pd
```

```python
dataset = pd.read_csv("insurance_pre.csv")
```

```python
dataset= pd.get_dummies(dataset, drop_first=True, dtype=int)
```

```python
dataset
```

```python
independent = dataset [['age', 'bmi', 'children','sex_male', 'smoker_yes']]
```

```python
dependent = dataset [['charges']]
```

```python
from lightgbm import LGBMRegressor
from sklearn.model_selection import GridSearchCV
param_grid = {'n_estimators':[50,60], 'objective':['regression', 'absolute error', 'logistic'], 'booster': ['gbtree', 'dart', 'gblinear']}
grid = GridSearchCV(LGBMRegressor(), param_grid, refit = True, verbose= 3,n_jobs=-1)
grid.fit(independent,dependent)
print("The R_score value for best parameter {}:".format(grid.best_params_),grid.best_score_)
```
