```python
import pandas as pd
```

```python
dataset = pd.read_csv("insurance_pre.csv")
```

```python
dataset= pd.get_dummies(dataset, drop_first=True, dtype=int)
```

```python
independent = dataset [['age', 'bmi', 'children','sex_male', 'smoker_yes']]
```

```python
dependent = dataset [['charges']]
```

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
from sklearn.model_selection import GridSearchCV
param_grid = {'criterion':['mse','friedman_mse', 'poisson', 'squared_error', 'absolute_error'],'max_features': ['auto','sqrt','log2'],'n_estimators':[10,100]}
grid = GridSearchCV(RandomForestRegressor(), param_grid, refit = True, verbose= 3,n_jobs=-1)
grid.fit(independent, dependent)
print("The R_score value for best parameter {}:".format(grid.best_params_),grid.best_score_)
```
